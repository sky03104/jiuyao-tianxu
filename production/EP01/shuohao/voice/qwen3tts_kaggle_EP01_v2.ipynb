```python
# 設計參考聲音（VoiceDesign）；T4 → float32＋sdpa
import torch, soundfile as sf, gc
from qwen_tts import Qwen3TTSModel
DTYPE = torch.float32
design = Qwen3TTSModel.from_pretrained("Qwen/Qwen3-TTS-12Hz-1.7B-VoiceDesign", device_map="cuda:0", dtype=DTYPE, attn_implementation="sdpa")
REFS = {}
for vid, v in VOICES.items():
    # 旁白的參考句也用旁白口吻，比較容易帶出史詩感
    text = "群山之上，雲海翻湧，一座古老的學院，靜靜矗立。" if vid == "VO" else REF_TEXT
    wavs, sr = design.generate_voice_design(text=text, language="Chinese", instruct=v["prompt_v2"])
    sf.write(f"{OUT}/ref/{vid}_{v['name']}.wav", wavs[0], sr); REFS[vid] = (wavs[0], sr, text)
    print("ref", vid, v["name"], f"{len(wavs[0])/sr:.1f}s")
del design; gc.collect(); torch.cuda.empty_cache()
```

```python
# 複製聲音唸台詞，每句 TAKES 個版本（Base）
clone = Qwen3TTSModel.from_pretrained("Qwen/Qwen3-TTS-12Hz-1.7B-Base", device_map="cuda:0", dtype=DTYPE, attn_implementation="sdpa")
PROMPTS = {vid: clone.create_voice_clone_prompt(ref_audio=(w, sr), ref_text=t) for vid, (w, sr, t) in REFS.items()}
REPORT = []
for l in LINES:
    for k in range(1, TAKES + 1):
        wavs, sr = clone.generate_voice_clone(text=l["text"], language="Chinese", voice_clone_prompt=PROMPTS[l["speaker"]])
        sf.write(f"{OUT}/clone/{l['id']}_{l['name']}_t{k}.wav", wavs[0], sr)
        REPORT.append(f"{l['id']}  {l['name']}  t{k}  {len(wavs[0])/sr:.2f}s / 分鏡 {l['cutDur']}s  {l['text']}")
print("\n".join(REPORT))
```

```python
import shutil
open(f"{OUT}/report.txt", "w").write("\n".join(REPORT))
shutil.make_archive("/kaggle/working/EP01_voice_v2", "zip", OUT)
print("OK → /kaggle/working/EP01_voice_v2.zip")
```
