```python
import torch, soundfile as sf
from qwen_tts import Qwen3TTSModel
clone = Qwen3TTSModel.from_pretrained("Qwen/Qwen3-TTS-12Hz-1.7B-Base", device_map="cuda:0", dtype=torch.float32, attn_implementation="sdpa")
PROMPTS = {}
for key, who, text in JOBS:
    if who not in PROMPTS:
        w, sr0 = sf.read(f"/tmp/ref/{who}.wav")
        PROMPTS[who] = clone.create_voice_clone_prompt(ref_audio=(w, sr0), ref_text=REFS[who]['text'])
    for k in (1, 2):
        try:
            wavs, sr = clone.generate_voice_clone(text=text, language="Chinese", voice_clone_prompt=PROMPTS[who], max_new_tokens=12 * 16)
        except TypeError:
            wavs, sr = clone.generate_voice_clone(text=text, language="Chinese", voice_clone_prompt=PROMPTS[who])
        sf.write(f"{OUT}/{key}_t{k}.wav", wavs[0][: int(sr * 16)], sr)
        print(key, k, f"{len(wavs[0])/sr:.2f}s", text)
```

```python
import glob
for w in sorted(glob.glob(f"{OUT}/*.wav")):
    m = w[:-4] + ".mp3"
    subprocess.run(["ffmpeg", "-y", "-loglevel", "error", "-i", w, "-ac", "1", "-b:a", "96k", m], check=True)
    print(f"@@MP3 {os.path.basename(m)} {base64.b64encode(open(m, 'rb').read()).decode()}")
```
