```python
import subprocess
print(subprocess.run("pip install -q -U qwen-tts soundfile 2>&1 | tail -3; nvidia-smi --query-gpu=name --format=csv", shell=True, capture_output=True, text=True).stdout)
```

```python
import json, os, base64, subprocess
REFS = json.loads(r'''{"裴含章": {"text": "新芽都在东廊。顺着左手边的回廊走到底就是。刚冒头，还没长。去年也有个孩子这样问我。", "b64": ""}, "厲若楓": {"text": "别急。他的脚步有点乱。它抬手之前，左肩会先沉。", "b64": ""}}''')
JOBS = json.loads(r'''[["S02-06_A03f2_裴含章", "裴含章", "补录的。最后一个。"], ["S02-40_A12f3_厲若楓", "厲若楓", "厉若枫。……走了。"]]''')
QI = json.loads(r'''{"ref_text": "今天天氣不錯，我們慢慢走過去吧，路上還可以再聊一聊。", "b64": ""}''')
OUT = '/kaggle/working/out'; os.makedirs(OUT, exist_ok=True); os.makedirs('/tmp/ref', exist_ok=True)
for n, r in REFS.items():
    open(f'/tmp/ref/{n}.mp3','wb').write(base64.b64decode(r['b64']))
    subprocess.run(['ffmpeg','-y','-loglevel','error','-i',f'/tmp/ref/{n}.mp3','-ar','24000',f'/tmp/ref/{n}.wav'],check=True)
open('/tmp/ref/qi.flac','wb').write(base64.b64decode(QI['b64']))
subprocess.run(['ffmpeg','-y','-loglevel','error','-i','/tmp/ref/qi.flac','-ar','24000','/tmp/ref/齊衡烈.wav'],check=True)
REFS['齊衡烈'] = {'text': QI['ref_text']}
```

```python
import torch, soundfile as sf
from qwen_tts import Qwen3TTSModel
clone = Qwen3TTSModel.from_pretrained("Qwen/Qwen3-TTS-12Hz-1.7B-Base", device_map="cuda:0", dtype=torch.float32, attn_implementation="sdpa")
PROMPTS = {}
for key, who, text in JOBS:
    if who not in PROMPTS:
        w, sr0 = sf.read(f"/tmp/ref/{who}.wav")
        PROMPTS[who] = clone.create_voice_clone_prompt(ref_audio=(w, sr0), ref_text=REFS[who]['text'])
    for k in (1, 2, 3):
        try:
            wavs, sr = clone.generate_voice_clone(text=text, language="Chinese", voice_clone_prompt=PROMPTS[who], max_new_tokens=12 * 16)
        except TypeError:
            wavs, sr = clone.generate_voice_clone(text=text, language="Chinese", voice_clone_prompt=PROMPTS[who])
        sf.write(f"{OUT}/{key}_t{k}.wav", wavs[0][: int(sr * 16)], sr)
        print(key, k, f"{len(wavs[0])/sr:.2f}s", text)
```

```python
import glob
for w in sorted(glob.glob(f"{OUT}/*.wav")):
    m = w[:-4] + ".mp3"
    subprocess.run(["ffmpeg", "-y", "-loglevel", "error", "-i", w, "-ac", "1", "-b:a", "96k", m], check=True)
    print(f"@@MP3 {os.path.basename(m)} {base64.b64encode(open(m, 'rb').read()).decode()}")
```
